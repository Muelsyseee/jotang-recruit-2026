```python
import torch
import numpy as np

# 分步骤前向传播
# f(x,y)=(xy+x**2)**2
x = 2.0
y = 3.0
a = x * y
b = x ** 2
c = a + b
f = c ** 2

print("a:", a)
print("b:", b)
print("c:", c)
print("f:", f)
```

# 计算图：
![计算图](./images/计算图.jpg)

```python
# 根据计算图计算梯度
grad_f = 1
grad_c = 2 * c
grad_a = grad_c
grad_b = grad_c
grad_x = grad_a * y + grad_b * 2 * x
grad_y = grad_a * y

print("grad_f:", grad_f)
print("grad_c:", grad_c)
print("grad_a:", grad_a)
print("grad_b:", grad_b)
print("grad_x:", grad_x)
print("grad_y:", grad_y)
```

```python
# autograd验证
x_tensor = torch.tensor(2.0, requires_grad=True)
y_tensor = torch.tensor(3.0, requires_grad=True)

a_tensor = x_tensor * y_tensor
b_tensor = x_tensor ** 2
c_tensor = a_tensor + b_tensor
f_tensor = c_tensor ** 2

f_tensor.backward()

print("grad_x_tensor:", x_tensor.grad.item())
print("grad_y_tensor:", y_tensor.grad.item())
```
