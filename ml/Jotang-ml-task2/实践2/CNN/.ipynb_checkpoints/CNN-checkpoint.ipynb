```python
from pathlib import Path
import random
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from sklearn.model_selection import train_test_split
import shutil
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from torch import nn
from torch import optim
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

repo_dir = next(
    path
    for path in (Path("."), Path(".."), Path("../.."), Path("../../.."))
    if (path / "实践1" / "outer1.png").is_file()
)

random_seed = 42
batch_size = 32
random.seed(random_seed)
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
learning_rate = 0.0003
epochs = 15
torch.manual_seed(random_seed)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(random_seed)
```

```python
# 统计数据

project_dir = repo_dir / "实践2" / "CNN"
image_dir = project_dir / "data" / "raw"

cat_dir = image_dir / "Cat"
dog_dir = image_dir / "Dog"

cats = list(cat_dir.glob("*.jpg"))
dogs = list(dog_dir.glob("*.jpg"))

cat_num = len(cats)
dog_num = len(dogs)

print("猫数量：", cat_num)
print("狗数量：", dog_num)
print("总数量：", cat_num + dog_num)
```

```python
# 随机展示数据
cat_sample = random.sample(cats, 5)
dog_sample = random.sample(dogs, 5)

fig, axes = plt.subplots(2, 5, figsize=(12, 6))

for row, samples in enumerate([cat_sample, dog_sample]):
    for col, path in enumerate(samples):
        with Image.open(path) as img:
            w, h = img.size
            rgb_image = img.convert("RGB")

            axes[row, col].imshow(rgb_image)
            axes[row, col].set_title(
                f"{'Cat' if row == 0 else 'dog'} ({w}*{h})"
            )
            axes[row, col].axis("off")

plt.show()
```

```python
# 划分数据集（70%:15%:15%）并存放
# ***不需要重新运行***
all_path = sorted(cats) + sorted(dogs)
all_labels = [0] * cat_num + [1] * dog_num

train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    all_path,
    all_labels,
    test_size=0.3,
    random_state=random_seed,
    stratify=all_labels
)

val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths,
    temp_labels,
    test_size=0.5,
    random_state=random_seed,
    stratify=temp_labels
)

split_dir = project_dir / "data" / "split"

split_data = {
    "train":(train_paths, train_labels),
    "val":(val_paths, val_labels),
    "test":(test_paths, test_labels)
}

for split_name in split_data:
    for class_name in ["Cat", "Dog"]:
        folder = split_dir / split_name / class_name
        folder.mkdir(parents=True, exist_ok=True)

for split_name, (path, label) in split_data.items():
    for src_path, label in zip(path, label):
        class_name = "Cat" if label == 0 else "Dog"
        dst_path = split_dir / split_name / class_name / src_path.name
        shutil.copy2(src_path, dst_path)

print("数据集已划分并存放")
```

```python
# 数据读取和预处理
split_dir = project_dir / "data" / "split"

image_transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

train_dataset = datasets.ImageFolder(
    root = split_dir / "train",
    transform = image_transform
)

val_dataset = datasets.ImageFolder(
    root = split_dir / "val",
    transform = image_transform
)

test_dataset = datasets.ImageFolder(
    root = split_dir / "test",
    transform = image_transform
)

train_num = len(train_dataset)
val_num = len(val_dataset)
test_num = len(test_dataset)

print("训练集数量：", train_num)
print("验证集数量：", val_num)
print("测试集数量：", test_num)
```

```python
# 预处理图像对比
index = random.randrange(len(train_dataset))
image_test, label_test = train_dataset[index]
image_test_path, _= train_dataset.samples[index]

with Image.open(image_test_path) as ima:
    image_test_origin = ima.convert("RGB")

print("原参数：")
print(image_test_origin.size)

print("处理后参数：")
print(image_test.shape)
print(label_test)
print(train_dataset.class_to_idx)

image_test_display = image_test.permute(1, 2, 0)
label_test_class = train_dataset.classes[label_test]

print("\n预处理图像比较")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(image_test_origin)
axes[0].set_title(f"{label_test_class} Before")
axes[0].axis("off")

axes[1].imshow(image_test_display)
axes[1].set_title(f"{label_test_class} After")
axes[1].axis("off")

plt.show()
```

```python
# 组织成batch迭代器
train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

images_test, labels_test= next(iter(train_loader))
print("batch中图像的shape：", images_test.shape)
print("batch中标签的shape：", labels_test.shape)
print("前十个标签：", labels_test[:10])
```

```python
# CNN网络搭建
'''
网络结构：
Input (3, 128, 128)
        │
Conv 3→16, 3×3 + ReLU
Conv 16→16, 3×3 + ReLU
MaxPool 2×2
        │
   (16, 64, 64)
        │
Conv 16→32, 3×3 + ReLU
Conv 32→32, 3×3 + ReLU
MaxPool 2×2
        │
   (32, 32, 32)
        │
Conv 32→64, 3×3 + ReLU
Conv 64→64, 3×3 + ReLU
MaxPool 2×2
        │
   (64, 16, 16)
        │
      Flatten
        │
 Linear 16384→128
        │
       ReLU
        │
   Linear 128→2
        │
  Cat / Dog logits
'''
class CNN(nn.Module):

    def __init__(self):
        super().__init__()
    
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),
    
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),
    
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),
        )
    
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64 * 16 * 16, 128),
            nn.ReLU(),
            nn.Dropout(p=0.3),
            nn.Linear(128, 2)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
        
model = CNN()
model
```

```python
# 记录图片经过网络的变化
x_test = images_test[:1]

model.eval()

with torch.no_grad():
    print("输入shape：", tuple(x_test.shape))

    for i, layer in enumerate(model.features):
        x_test = layer(x_test)

        print(
            f"features层{i}:{type(layer).__name__},输出shape:{tuple(x_test.shape)}"
        )

    for i, layer in enumerate(model.classifier):
        x_test = layer(x_test)

        print(
            f"classifier层{i}:{type(layer).__name__},输出shape:{tuple(x_test.shape)}"
        )
```

```python
# 模型训练参数
model = model.to(device)
print("训练设备：", device)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate,
    # weight_decay=0.0001
)

print("损失函数：", criterion)
print("优化器：", optimizer.__class__.__name__)
print("学习率：", learning_rate)
```

```python
# 模型训练与保存
save_dir = project_dir / "models"
save_dir.mkdir(parents=True, exist_ok=True)

model_path = save_dir / "model.pth"

train_loss = []
train_accuracy = []
eval_loss = []
eval_accuracy = []

best_accuracy = -1.0
best_loss = float("inf")
best_epoch = 0

for epoch in range(epochs):
    model.train()
    
    total_loss_train = 0.0
    total_correct_train = 0
    total_samples_train = 0
    
    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)
    
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    
        current_batch_size = labels.size(0)
        total_samples_train += current_batch_size
        total_loss_train += loss.item() * current_batch_size
        predictions = outputs.argmax(dim=1)
        correct = (predictions == labels).sum().item()
        total_correct_train += correct

    train_loss_epoch = total_loss_train / total_samples_train
    train_accuracy_epoch = total_correct_train / total_samples_train

    train_loss.append(train_loss_epoch)
    train_accuracy.append(train_accuracy_epoch)


    model.eval()

    total_loss_eval = 0.0
    total_correct_eval = 0
    total_samples_eval = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)
        
            outputs = model(images)
            loss = criterion(outputs, labels)

            current_batch_size = labels.size(0)
            total_samples_eval += current_batch_size
            total_loss_eval += loss.item() * current_batch_size
            predictions = outputs.argmax(dim=1)
            correct = (predictions == labels).sum().item()
            total_correct_eval += correct

    eval_loss_epoch = total_loss_eval / total_samples_eval
    eval_accuracy_epoch = total_correct_eval / total_samples_eval

    eval_accuracy.append(eval_accuracy_epoch)
    eval_loss.append(eval_loss_epoch)

    is_best = (
        eval_accuracy_epoch > best_accuracy
        or (
            eval_accuracy_epoch == best_accuracy
            and eval_loss_epoch < best_loss
        )
    )

    if is_best:
        best_epoch = epoch + 1
        best_accuracy = eval_accuracy_epoch
        best_loss = eval_loss_epoch

        torch.save(
            model.state_dict(),
            model_path
        )

        print(f"已更新模型参数:{model_path}")

    print(f"epoch{epoch + 1}:\ntrain_accuracy:{train_accuracy_epoch}, train_loss:{train_loss_epoch}\neval_accuracy:{eval_accuracy_epoch}, eval_loss:{eval_loss_epoch}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
epochs_plot = range(1, epochs + 1)

axes[0].plot(epochs_plot, train_accuracy, marker="o", label="Train Accuracy")
axes[0].plot(epochs_plot, eval_accuracy, marker="o", label="Eval Accuracy")
axes[0].set_title("Train/Eval Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].axvline(
    best_epoch, linestyle="--", label="Best Epoch"
)
axes[0].legend()
axes[0].grid(True)


axes[1].plot(epochs_plot, train_loss, marker="o", label="Train Loss")
axes[1].plot(epochs_plot, eval_loss, marker="o", label="Eval Loss")
axes[1].set_title("Train/Eval Loss")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].axvline(
    best_epoch, linestyle="--", label="Best Epoch"
)
axes[1].legend()
axes[1].grid(True)


figure_dir = project_dir / "figures"
figure_dir.mkdir(parents=True, exist_ok=True)
figure_path = figure_dir / "accuracy_and_loss.png"

fig.savefig(figure_path, dpi=150, bbox_inches="tight")

plt.show()

print(f"{epochs}轮训练已完成")
print(f"最佳轮次:{best_epoch}")
print(f"最佳accuracy:{best_accuracy}")
print(f"最佳loss:{best_loss}")
print(f"已保存模型路径:{model_path}")
print(f"已保存图像路径：{figure_path}")
```

```python
# 加载最佳模型并测试
test_model = CNN().to(device)

model_path = project_dir / "models" / "model_best.pth"
weights = torch.load(
    model_path,
    map_location=device,
    weights_only=True
)
test_model.load_state_dict(weights)

test_model.eval()

total_loss_test = 0.0
total_correct_test = 0
total_samples_test = 0

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)

        outputs = test_model(images)
        loss = criterion(outputs, labels)

        batch_size_now = labels.size(0)

        total_loss_test += loss.item() * batch_size_now
        total_samples_test += batch_size_now

        predictions = outputs.argmax(dim=1)
        total_correct_test += (
            (predictions == labels).sum().item()
        )

test_loss = total_loss_test / total_samples_test
test_accuracy = total_correct_test / total_samples_test

print(f"测试图片数量：{total_samples_test}")
print(f"Test Loss：{test_loss}")
print(f"Test Accuracy：{test_accuracy}")
```

```python
# 数据增强实验，重新准备dataloader
enhance_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        128,
        scale=(0.8, 1.0)
    ),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(
        brightness=0.5,
        contrast=0.5,
        saturation=0.2
    ),
    transforms.ToTensor()
])

train_enhance_dataset = datasets.ImageFolder(
    root=split_dir / "train",
    transform=enhance_transform
)

train_enhance_loader = DataLoader(
    train_enhance_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

train_enhance_num = len(train_enhance_dataset)
print("增强训练集数量：", train_enhance_num)
```

```python
# 数据增强演示
index = random.randrange(len(train_enhance_dataset))
image_path, label = train_enhance_dataset.samples[index]

with Image.open(image_path) as img:
    original_image = img.convert("RGB")

fig, axes = plt.subplots(1, 5, figsize=(16, 4))

axes[0].imshow(original_image)
axes[0].set_title("Original")
axes[0].axis("off")

for i in range(1, 5):
    enhance_image = enhance_transform(original_image)
    enhance_image_display = enhance_image.permute(1, 2, 0)

    axes[i].imshow(enhance_image_display)
    axes[i].set_title(f"Enhanced {i}")
    axes[i].axis("off")

plt.show()

print("真实类别：", train_enhance_dataset.classes[label])
```

```python
# 数据增强对应模型和优化器创建
model_enhance = CNN().to(device)

optimizer_enhance = torch.optim.Adam(
    model_enhance.parameters(),
    lr=learning_rate
)
```

```python
# 数据增强模型训练与保存
save_dir = project_dir / "models"
model_path_enhance = save_dir / "model_enhance.pth"

train_loss_enhance = []
train_accuracy_enhance = []
eval_loss_enhance = []
eval_accuracy_enhance = []

best_accuracy_enhance = -1.0
best_loss_enhance = float("inf")
best_epoch_enhance = 0

for epoch in range(epochs):
    model_enhance.train()

    total_loss_train_enhance = 0.0
    total_correct_train_enhance = 0
    total_samples_train_enhance = 0

    for images, labels in train_enhance_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer_enhance.zero_grad()
        outputs = model_enhance(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer_enhance.step()

        current_batch_size = labels.size(0)
        total_samples_train_enhance += current_batch_size
        total_loss_train_enhance += loss.item() * current_batch_size

        predictions = outputs.argmax(dim=1)
        correct = (predictions == labels).sum().item()
        total_correct_train_enhance += correct

    train_loss_epoch_enhance = total_loss_train_enhance / total_samples_train_enhance
    train_accuracy_epoch_enhance = total_correct_train_enhance / total_samples_train_enhance

    train_loss_enhance.append(train_loss_epoch_enhance)
    train_accuracy_enhance.append(train_accuracy_epoch_enhance)

    model_enhance.eval()

    total_loss_eval_enhance = 0.0
    total_correct_eval_enhance = 0
    total_samples_eval_enhance = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model_enhance(images)
            loss = criterion(outputs, labels)

            current_batch_size = labels.size(0)
            total_samples_eval_enhance += current_batch_size
            total_loss_eval_enhance += loss.item() * current_batch_size

            predictions = outputs.argmax(dim=1)
            correct = (predictions == labels).sum().item()
            total_correct_eval_enhance += correct

    eval_loss_epoch_enhance = total_loss_eval_enhance / total_samples_eval_enhance
    eval_accuracy_epoch_enhance = total_correct_eval_enhance / total_samples_eval_enhance

    eval_accuracy_enhance.append(eval_accuracy_epoch_enhance)
    eval_loss_enhance.append(eval_loss_epoch_enhance)

    is_best_enhance = (
        eval_accuracy_epoch_enhance > best_accuracy_enhance
        or (
            eval_accuracy_epoch_enhance == best_accuracy_enhance
            and eval_loss_epoch_enhance < best_loss_enhance
        )
    )

    if is_best_enhance:
        best_epoch_enhance = epoch + 1
        best_accuracy_enhance = eval_accuracy_epoch_enhance
        best_loss_enhance = eval_loss_epoch_enhance

        torch.save(
            model_enhance.state_dict(),
            model_path_enhance
        )

        print(f"已更新模型参数:{model_path_enhance}")

    print(
        f"epoch{epoch + 1}:\n"
        f"train_accuracy:{train_accuracy_epoch_enhance}, "
        f"train_loss:{train_loss_epoch_enhance}\n"
        f"eval_accuracy:{eval_accuracy_epoch_enhance}, "
        f"eval_loss:{eval_loss_epoch_enhance}"
    )

fig_enhance, axes_enhance = plt.subplots(1, 2, figsize=(12, 5))
epochs_plot_enhance = range(1, epochs + 1)

axes_enhance[0].plot(
    epochs_plot_enhance, train_accuracy_enhance,
    marker="o", label="Train Accuracy"
)
axes_enhance[0].plot(
    epochs_plot_enhance, eval_accuracy_enhance,
    marker="o", label="Eval Accuracy"
)
axes_enhance[0].set_title("Train/Eval Accuracy (Enhance)")
axes_enhance[0].set_xlabel("Epoch")
axes_enhance[0].set_ylabel("Accuracy")
axes_enhance[0].axvline(
    best_epoch_enhance, linestyle="--", label="Best Epoch"
)
axes_enhance[0].legend()
axes_enhance[0].grid(True)


axes_enhance[1].plot(
    epochs_plot_enhance, train_loss_enhance,
    marker="o", label="Train Loss"
)
axes_enhance[1].plot(
    epochs_plot_enhance, eval_loss_enhance,
    marker="o", label="Eval Loss"
)
axes_enhance[1].set_title("Train/Eval Loss (Enhance)")
axes_enhance[1].set_xlabel("Epoch")
axes_enhance[1].set_ylabel("Loss")
axes_enhance[1].axvline(
    best_epoch_enhance, linestyle="--", label="Best Epoch"
)
axes_enhance[1].legend()
axes_enhance[1].grid(True)

figure_dir = project_dir / "figures"
figure_path_enhance = figure_dir / "accuracy_and_loss_enhance.png"

fig_enhance.savefig(
    figure_path_enhance,
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(f"{epochs}轮训练已完成")
print(f"最佳轮次:{best_epoch_enhance}")
print(f"最佳accuracy:{best_accuracy_enhance}")
print(f"最佳loss:{best_loss_enhance}")
print(f"已保存模型路径:{model_path_enhance}")
print(f"已保存图像路径：{figure_path_enhance}")
```

```python
# 错误样本分析
analysis_model = CNN().to(device)

analysis_model.load_state_dict(
    torch.load(
        project_dir / "models" / "model_best.pth",
        map_location=device,
        weights_only=True
    )
)

analysis_model.eval()

analysis_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0
)

wrong_samples = []
sample_offset = 0

with torch.no_grad():

    for images, labels in analysis_loader:
        images = images.to(device)

        outputs = analysis_model(images)
        probabilities = torch.softmax(outputs, dim=1)
        confidence, predictions = probabilities.max(dim=1)

        predictions = predictions.cpu()
        confidence = confidence.cpu()

        for i in range(len(labels)):
            true_label = labels[i].item()
            pred_label = predictions[i].item()
            
            if pred_label != true_label:

                dataset_index = sample_offset + i
                image_path, _ = val_dataset.samples[dataset_index]

                wrong_samples.append({
                    "path": image_path,
                    "true": true_label,
                    "pred": pred_label,
                    "confidence": confidence[i].item()
                })

        sample_offset += len(labels)

print("验证集样本数：", len(val_dataset))
print("分类错误数量：", len(wrong_samples))
print(f"错误率：{len(wrong_samples) / len(val_dataset)}")
```

```python
# 错误样本可视化
num_display = min(8, len(wrong_samples))

random.seed(42)
samples = random.sample(wrong_samples, num_display)

if num_display == 0:
    print("当前模型在验证集上没有发现分类错误样本。")

else:
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))

    axes = axes.flatten()

    for i, sample in enumerate(samples):

        with Image.open(sample["path"]) as img:
            image = img.convert("RGB")

        true_name = val_dataset.classes[sample["true"]]
        pred_name = val_dataset.classes[sample["pred"]]

        axes[i].imshow(image)

        axes[i].set_title(
            f"True: {true_name} | Pred: {pred_name}\n"
            f"Confidence: {sample['confidence']:.2%}"
        )

        axes[i].axis("off")

    for i in range(num_display, len(axes)):
        axes[i].axis("off")

    fig.tight_layout()
    plt.show()
```

```python
# 输出混淆矩阵
all_true = []
all_pred = []

analysis_model.eval()

with torch.no_grad():
    for images, labels in analysis_loader:
        images = images.to(device)

        outputs = analysis_model(images)
        predictions = outputs.argmax(dim=1)

        # API：Tensor.tolist()
        # 将 Tensor 转换为 Python 列表
        # 示例：torch.tensor([0, 1]).tolist() -> [0, 1]
        all_true.extend(labels.tolist())
        all_pred.extend(predictions.cpu().tolist())

cm = confusion_matrix(
    all_true,
    all_pred,
    labels=[0, 1]
)

print("混淆矩阵：")
print(cm)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=val_dataset.classes
)

fig, ax = plt.subplots(figsize=(6, 5))

display.plot(ax=ax, cmap="Blues", values_format="d")

ax.set_title("Validation Confusion Matrix")
plt.show()
```

```python
# 新预处理dataloader准备
# 1. 等比例缩放 + Padding
pad_transform_new = transforms.Lambda(
    lambda img: ImageOps.pad(
        img,
        size=(128, 128),
        method=Image.Resampling.BILINEAR,
        color=(128, 128, 128)
    )
)

# 2. 训练集：保留数据增强，但不使用 RandomResizedCrop
image_transform_new = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(
        brightness=0.5,
        contrast=0.5,
        saturation=0.2
    ),
    pad_transform_new,
    transforms.ToTensor()
])

# 3. 验证集、测试集：仅使用确定性预处理
eval_transform_new = transforms.Compose([
    pad_transform_new,
    transforms.ToTensor()
])

# 4. Dataset
train_dataset_new = datasets.ImageFolder(
    split_dir / "train",
    transform=image_transform_new
)

val_dataset_new = datasets.ImageFolder(
    split_dir / "val",
    transform=eval_transform_new
)

test_dataset_new = datasets.ImageFolder(
    split_dir / "test",
    transform=eval_transform_new
)

# 5. DataLoader
batch_size_new = batch_size

train_loader_new = DataLoader(
    train_dataset_new,
    batch_size=batch_size_new,
    shuffle=True,
    num_workers=0
)

eval_loader_new = DataLoader(
    val_dataset_new,
    batch_size=batch_size_new,
    shuffle=False,
    num_workers=0
)

test_loader_new = DataLoader(
    test_dataset_new,
    batch_size=batch_size_new,
    shuffle=False,
    num_workers=0
)

print("Train:", len(train_dataset_new))
print("Val:", len(val_dataset_new))
print("Test:", len(test_dataset_new))
print("类别映射：", train_dataset_new.class_to_idx)
```

```python
# 新预处理模型训练验证
model_new = CNN().to(device)

optimizer_new = optim.Adam(
    model_new.parameters(),
    lr=learning_rate
)

save_dir_new = project_dir / "models"
save_dir_new.mkdir(parents=True, exist_ok=True)

model_path_new = save_dir_new / "model_new.pth"

train_loss_new = []
train_accuracy_new = []
eval_loss_new = []
eval_accuracy_new = []

best_accuracy_new = -1.0
best_loss_new = float("inf")
best_epoch_new = 0


for epoch in range(epochs):
    model_new.train()

    total_loss_train_new = 0.0
    total_correct_train_new = 0
    total_samples_train_new = 0

    for images, labels in train_loader_new:

        images = images.to(device)
        labels = labels.to(device)

        optimizer_new.zero_grad()

        outputs = model_new(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer_new.step()

        current_batch_size = labels.size(0)

        total_samples_train_new += current_batch_size
        total_loss_train_new += loss.item() * current_batch_size

        predictions = outputs.argmax(dim=1)
        correct = (predictions == labels).sum().item()

        total_correct_train_new += correct

    train_loss_epoch_new = (
        total_loss_train_new / total_samples_train_new
    )

    train_accuracy_epoch_new = (
        total_correct_train_new / total_samples_train_new
    )

    train_loss_new.append(train_loss_epoch_new)
    train_accuracy_new.append(train_accuracy_epoch_new)

    model_new.eval()

    total_loss_eval_new = 0.0
    total_correct_eval_new = 0
    total_samples_eval_new = 0

    with torch.no_grad():

        for images, labels in eval_loader_new:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model_new(images)
            loss = criterion(outputs, labels)

            current_batch_size = labels.size(0)

            total_samples_eval_new += current_batch_size
            total_loss_eval_new += loss.item() * current_batch_size

            predictions = outputs.argmax(dim=1)
            correct = (predictions == labels).sum().item()

            total_correct_eval_new += correct

    eval_loss_epoch_new = (
        total_loss_eval_new / total_samples_eval_new
    )

    eval_accuracy_epoch_new = (
        total_correct_eval_new / total_samples_eval_new
    )

    eval_accuracy_new.append(eval_accuracy_epoch_new)
    eval_loss_new.append(eval_loss_epoch_new)


    is_best_new = (
        eval_accuracy_epoch_new > best_accuracy_new
        or (
            eval_accuracy_epoch_new == best_accuracy_new
            and eval_loss_epoch_new < best_loss_new
        )
    )

    if is_best_new:

        best_epoch_new = epoch + 1
        best_accuracy_new = eval_accuracy_epoch_new
        best_loss_new = eval_loss_epoch_new

        torch.save(
            model_new.state_dict(),
            model_path_new
        )

        print(f"已更新模型参数：{model_path_new}")


    print(
        f"epoch {epoch + 1}:\n"
        f"train_accuracy: {train_accuracy_epoch_new:.4f}, "
        f"train_loss: {train_loss_epoch_new:.4f}\n"
        f"eval_accuracy: {eval_accuracy_epoch_new:.4f}, "
        f"eval_loss: {eval_loss_epoch_new:.4f}"
    )

fig_new, axes_new = plt.subplots(1, 2, figsize=(12, 5))

epochs_plot_new = range(1, len(train_loss_new) + 1)


axes_new[0].plot(
    epochs_plot_new,
    train_accuracy_new,
    marker="o",
    label="Train Accuracy"
)

axes_new[0].plot(
    epochs_plot_new,
    eval_accuracy_new,
    marker="o",
    label="Eval Accuracy"
)

axes_new[0].set_title("Train/Eval Accuracy (New)")
axes_new[0].set_xlabel("Epoch")
axes_new[0].set_ylabel("Accuracy")

axes_new[0].axvline(
    best_epoch_new,
    linestyle="--",
    label="Best Epoch"
)

axes_new[0].legend()
axes_new[0].grid(True)

axes_new[1].plot(
    epochs_plot_new,
    train_loss_new,
    marker="o",
    label="Train Loss"
)

axes_new[1].plot(
    epochs_plot_new,
    eval_loss_new,
    marker="o",
    label="Eval Loss"
)

axes_new[1].set_title("Train/Eval Loss (New)")
axes_new[1].set_xlabel("Epoch")
axes_new[1].set_ylabel("Loss")

axes_new[1].axvline(
    best_epoch_new,
    linestyle="--",
    label="Best Epoch"
)

axes_new[1].legend()
axes_new[1].grid(True)

fig_new.tight_layout()

figure_dir_new = project_dir / "figures"
figure_dir_new.mkdir(parents=True, exist_ok=True)

figure_path_new = (
    figure_dir_new / "accuracy_and_loss_new.png"
)

fig_new.savefig(
    figure_path_new,
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(f"{len(train_loss_new)} 轮训练已完成")
print(f"最佳轮次：{best_epoch_new}")
print(f"最佳 accuracy：{best_accuracy_new:.4f}")
print(f"最佳 loss：{best_loss_new:.4f}")
print(f"已保存模型路径：{model_path_new}")
print(f"已保存图像路径：{figure_path_new}")
```
